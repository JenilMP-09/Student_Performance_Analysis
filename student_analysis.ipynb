# 3. Analysis Questions:

```python
stud_df.head()
```

* <font size="4">**Average final grade (G3):**</font>

```python
# Print average of final grade (G3), rounded to 2 decimals:
#
# - stud_df['G3'].mean() --> calculates the mean of the final grade (G3) column.
#
# - round(_, 2) --> rounds the result to 2 decimal places.

print(f"Average final grade will be: {round(stud_df["G3"].mean(), 2)}")
```

* <font size="4">**How many students scored above 15?**</font>

```python
# Print count of students with final grade (G3) above 15:
#
# - stud_df['G3'] > 15 --> filters rows where the final grade is greater than 15.
#
# - count() --> counts how many students meet that condition.

print(f"Total number of students who scored above 15 are: {stud_df["G3"][stud_df["G3"] > 15].count()}")
```

* <font size="4">**Is study time correlated with performance?**</font>

```python
print("Yes, study time is correlated with performance (final grade 'G3')")

# Calculate correlation between studytime and final grade (G3):
#
# - corr() --> computes the Pearson correlation coefficient between two columns.
#
# - Returns a value between -1 and 1:
# - Positive --> higher studytime tends to increase final grade.
# - Negative --> higher studytime tends to decrease final grade.
# - Near 0 --> little to no linear relationship.

print(f"Correlation between studytime and final grade (G3): ", stud_df["studytime"].corr(stud_df["G3"]))
```

* <font size="4">**Which gender performs better on average?**</font>

```python
# Calculate an average final grade by gender:

stud_df.groupby("sex")["G3"].mean()
```

* Based on above data "Male" performs better on average.

# 4. Visualizations:

* <font size="4">**Histogram of grades:**</font>

```python
# Plot histogram of final grades with kde:

plt.figure(figsize=(8, 6))
sns.histplot(stud_df["G3"], bins=10, kde=True)
plt.title("Histogram of Grades")
plt.xlabel("Grade")
plt.ylabel("Frequency")
plt.show()
```

* <font size="4">**Scatterplot: study time vs grades:**</font>

```python
# Scatter plot showing relationship between study time and grades:

plt.figure(figsize=(8, 6))
sns.scatterplot(x="studytime", y="G3", data=stud_df)
plt.title("Study Time vs Grades")
plt.xlabel("Study Time")
plt.ylabel("Grade")
plt.show()
```

* <font size="4">**Bar chart: male vs female average score:**</font>

```python
# Bar chart comparing an average grades by gender:

plt.figure(figsize=(8, 6))
sns.barplot(x="sex", y="G3", data=stud_df)
plt.title("Male vs Female Average Score")
plt.xlabel("Gender")
plt.ylabel("Average Grade")
plt.show()
```
